```python
import pandas as pd

input_file = "echo-170826/mixes_template.csv"
output_file = "echo-170826/mixes_template-1.csv"

# 1. Read the CSV file
df = pd.read_csv(input_file)

# 2. Overwrite the first column with sequentially numbered Mix labels
df.iloc[:, 0] = [f"Mix no{i}" for i in range(1, len(df) + 1)]

# 3. Save to a new CSV file
df.to_csv(output_file, index=False)

print(f"Done! Processed {len(df)} rows.")
```

```python
import pandas as pd

input_file = "echo-140826/mixes_template.csv"
output_file = "echo-140826/mixes_template-1.csv"

# 1. Read the original CSV
df = pd.read_csv(input_file)

# 2. Insert new column at index 2 (after 2nd column) with header '0.0' and values 0.0
df.insert(loc=2, column="0.0", value=0.0)

# 3. Save to output
df.to_csv(output_file, index=False)

print("Done! Added new column '0.0' after the 2nd column.")
```
